```python
from mecab import MeCab

mecab = MeCab()

def morpheme(text):
    result = mecab.pos(text)
    for morph, tag in result:
        print(morph, tag)
```

```python
morpheme("선물용으로 빨리 받아서 전달했어야 하는 상품이었는데 머그컵만 와서 당황했습니다.")
```

```python
import hgtk

def tokenized(text):
    tokenized_data = []
    dec_words = ""
    dec_words_list = []
    tempX = mecab.morphs(text)
    tempX = [word for word in tempX]
    tokenized_data.append(tempX)

    for words in tokenized_data:
        for word in words:
            for i in word:
                if hgtk.checker.is_hangul(i) or i == ".":
                    if hgtk.checker.is_hangul(i):
                        dec_word = hgtk.letter.decompose(i)
                        for j in dec_word:
                            if j == "":
                                j = "-"
                            dec_words = dec_words + j
                    else:
                        dec_words = dec_words + i
            dec_words_list.append(dec_words)
            dec_words = ""
        print(dec_words_list)
        
```

```python
tokenized("선물용으로 빨리 받아서 전달했어야 하는 상품이었는데 머그컵만 와서 당황했습니다.")
```

```python
def compose(char):
    word = []
    words = []
    result = [char[i:i+3] for i in range(0, len(char), 3)]
    for char in result:
        print(char)
        for j in char:
            word.append(j)
            print(word)
        words = hgtk.letter.compose(word)
        word = []
    print(words)
```
